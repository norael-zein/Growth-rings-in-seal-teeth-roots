## Load data and import packages

```python
sys.path.append("../dataloaders")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision.models as models
from torchvision.models import ResNet18_Weights
from data_loader_cropping_padding import CroppedPaddedDataset
from data_loader_cropping_nopadding import CroppedDataset
from torch.utils.data import random_split, DataLoader
import matplotlib.pyplot as plt
import sys
import torch.nn.functional as F
import numpy as np
from scipy.signal import find_peaks
```

```python
#Load dataset

#Dataset with padding
dataset = CroppedPaddedDataset("../labels.csv", "../images")
#dataset = CroppedDataset("../labels.csv", "../images")

#Dataset without padding
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

## ResNet18 model

```python
class ResNetRegressor(nn.Module):
    def __init__(self):
        super().__init__()

        self.backbone = models.resnet18(weights=ResNet18_Weights.DEFAULT)

        # regression-head
        in_features = self.backbone.fc.in_features
        self.backbone.fc = nn.Linear(in_features, 1)

        #Freeze First layers
        for p in self.backbone.conv1.parameters():
            p.requires_grad = False
        for p in self.backbone.bn1.parameters():
            p.requires_grad = False
        #for p in self.backbone.layer1.parameters():
        #    p.requires_grad = False
        for p in self.backbone.layer2.parameters():
            p.requires_grad = False
        for p in self.backbone.layer3.parameters():
            p.requires_grad = False
        #for p in self.backbone.layer4.parameters():
        #    p.requires_grad = False
        #for p in self.backbone.fc.parameters():
        #    p.requires_grad = False

    def forward(self, x):
        return self.backbone(x)

model_ResNet = ResNetRegressor().to(device)
```

```python
#Training and validation data
val_split = 0.2  
n_total = len(dataset)
n_val = int(n_total * val_split)
n_train = n_total - n_val

train_set, val_set = random_split(dataset, [n_train, n_val])

train_loader = DataLoader(train_set, batch_size=32, shuffle=True)
val_loader = DataLoader(val_set, batch_size=32, shuffle=False)
```

```python
criterion = nn.SmoothL1Loss(beta=1.0)  #Huber-loss

optimizer = optim.Adam(model_ResNet.parameters(), lr=1e-4, weight_decay=1e-4)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=5)
num_epochs = 20

train_mse_list = []
val_mse_list = []
val_mae_list = []
```

```python
model_ResNet.train()

for epoch in range(num_epochs):
    #Train
    model_ResNet.train()
    running_mse = 0.0
    n_train = 0

    for imgs, labels, names in train_loader:
        imgs = imgs.to(device)
        labels = labels.float().view(-1, 1).to(device)

        optimizer.zero_grad()
        outputs = model_ResNet(imgs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        mse_batch = torch.mean((outputs - labels)**2).item()
        running_mse += mse_batch * imgs.size(0)
        n_train += imgs.size(0)

    train_mse_epoch = running_mse / n_train
    train_mse_list.append(train_mse_epoch)

    #Validation
    model_ResNet.eval()
    val_mse = 0.0
    val_mae = 0.0
    n_val = 0

    with torch.no_grad():
        for imgs, labels, names in val_loader:
            imgs = imgs.to(device)
            labels = labels.float().view(-1, 1).to(device)

            outputs = model_ResNet(imgs)
            mse_batch = torch.mean((outputs - labels)**2).item()
            mae_batch = torch.mean(torch.abs(outputs - labels)).item()

            val_mse += mse_batch * imgs.size(0)
            val_mae += mae_batch * imgs.size(0)
            n_val += imgs.size(0)

    val_mse_epoch = val_mse / n_val
    val_mae_epoch = val_mae / n_val

    val_mse_list.append(val_mse_epoch)
    val_mae_list.append(val_mae_epoch)

    scheduler.step(val_mae_epoch)

    print(f"Epoch {epoch+1}/{num_epochs} - "
          f"Train MSE: {train_mse_epoch:.3f} - "
          f"Val MSE: {val_mse_epoch:.3f} - "
          f"Val MAE: {val_mae_epoch:.3f}")
```

```python
#MSE
plt.figure()
plt.plot(train_mse_list, label="Train MSE")
plt.plot(val_mse_list, label="Val MSE")
plt.xlabel("Epoch")
plt.ylabel("Error")
plt.title("Training and Validation MSE (ResNet)")
plt.legend()
plt.show()

#MAE
plt.figure()
plt.plot(val_mae_list, label="Val MAE")
plt.xlabel("Epoch")
plt.ylabel("Error")
plt.title("Validation MAE (ResNet)")
plt.legend()
plt.show()
```

## Grad-CAM

```python
class GradCAM:
    def __init__(self, model, target_layer):
        self.model = model
        self.model.eval()
        self.target_layer = target_layer

        self.activations = None
        self.gradients = None

        #Save activations
        def forward_hook(module, inp, out):
            self.activations = out.detach()

        #Save gradients
        def backward_hook(module, grad_input, grad_output):
            self.gradients = grad_output[0].detach()

        self.fwd_handle = target_layer.register_forward_hook(forward_hook)
        self.bwd_handle = target_layer.register_full_backward_hook(backward_hook)

    def __call__(self, x):
        self.model.zero_grad()

        output = self.model(x)  
        target = output.squeeze()         
        target.backward(retain_graph=True)


        grads = self.gradients
        acts = self.activations

        weights = grads.mean(dim=(2, 3), keepdim=True)  
        cam = (weights * acts).sum(dim=1, keepdim=True)  
        cam = F.relu(cam)

        cam = F.interpolate(
            cam,
            size=x.shape[2:],
            mode="bilinear",
            align_corners=False
        )

        cam_min, cam_max = cam.min(), cam.max()
        if cam_max > cam_min:
            cam = (cam - cam_min) / (cam_max - cam_min + 1e-8)
        else:
            cam = torch.zeros_like(cam)
        return cam, output.detach()
```

### Grad-CAM on ResNet 18

```python
model_ResNet.eval()

#Determine layer ResNet18
target_layer_resnet = model_ResNet.backbone.layer1[-1].conv2
gradcam_resnet = GradCAM(model_ResNet, target_layer_resnet)

for idx in range(200):
    img, label, name = dataset[idx]
    img = img.unsqueeze(0).to(device)

    #Grad-CAM
    cam_resnet, pred_resnet = gradcam_resnet(img)

    print(f"\nImage {idx}")
    print("Filename:", name)
    print("True age:", label)
    print("Pred age:", pred_resnet.item())

    img_np = img[0].detach().cpu().permute(1, 2, 0).numpy()
    cam_np = cam_resnet[0, 0].detach().cpu().numpy()

    #Plot figure
    fig, ax = plt.subplots(1, 2, figsize=(8, 4))
    
    ax[0].imshow(img_np)
    ax[0].set_title("Original")
    ax[0].axis("off")

    ax[1].imshow(img_np, alpha=0.5)
    ax[1].imshow(cam_np, alpha=0.5, cmap="jet")
    ax[1].set_title("ResNet Grad-CAM")
    ax[1].axis("off")

    plt.tight_layout()
    plt.show()
```

### Grad-CAM Overlay on images

```python
model_ResNet.eval()

#Determine layer ResNet18
target_layer_resnet = model_ResNet.backbone.layer1[-1].conv2
gradcam_resnet = GradCAM(model_ResNet, target_layer_resnet)

for idx in range(200):
    img, label, name = dataset[idx]
    img = img.unsqueeze(0).to(device)

    #Grad-CAM
    cam_resnet, pred_resnet = gradcam_resnet(img)
    print(f"\nImage {idx}")
    print("Filename:", name)
    print("True age:", label)
    print("Pred age:", pred_resnet.item())

    #Convert to numpy
    img_np = img[0].detach().cpu().permute(1, 2, 0).numpy()
    cam_np = cam_resnet[0, 0].detach().cpu().numpy()

    #Normalize CAM to [0,1]
    cam_norm = (cam_np - cam_np.min()) / (cam_np.max() - cam_np.min() + 1e-8)

    #Only show the 30% strongest pixel values in heatmap
    threshold = np.quantile(cam_norm, 0.70)
    mask = cam_norm >= threshold   
    cam_masked = np.zeros_like(cam_norm)
    cam_masked[mask] = cam_norm[mask]

    heatmap = plt.cm.jet(cam_norm)[..., :3]

    #Overlay on original image
    overlay = 0.6 * img_np + 0.4 * heatmap

    #Plot
    fig, ax = plt.subplots(1, 2, figsize=(10, 4))

    ax[0].imshow(img_np)
    ax[0].set_title("Original")
    ax[0].axis("off")

    ax[1].imshow(overlay)
    ax[1].set_title("Grad-CAM Overlay (ResNet)")
    ax[1].axis("off")

    plt.tight_layout()
    plt.show()
```

### Grad-CAM point out

```python
model_ResNet.eval()

target_layer = model_ResNet.backbone.layer1[-1].conv2
gradcam = GradCAM(model_ResNet, target_layer)

for idx in range(300): 
    img, label, name = dataset[idx]
    img = img.unsqueeze(0).to(device)

    #Grad-CAM
    cam, pred = gradcam(img)
    pred_age = int(round(pred.item()))
    print("\nImage:", idx)
    print("Filename:", name)
    print("True age:", label)
    print("Pred age:", pred_age)
    img_np = img[0].detach().cpu().permute(1, 2, 0).numpy()
    cam_np = cam[0, 0].detach().cpu().numpy()

    #normalize
    img_np = (img_np - img_np.min()) / (img_np.max() - img_np.min() + 1e-8)
    cam_norm = (cam_np - cam_np.min()) / (cam_np.max() - cam_np.min() + 1e-8)

    #Horisontal line
    h, w = cam_norm.shape
    row = h // 2
    signal = cam_norm[row, :]

    ##Smooth
    kernel = np.ones(7) / 7.0
    signal_smooth = np.convolve(signal, kernel, mode="same")

    #Find peaks
    peaks, _ = find_peaks(signal_smooth, distance=4)
    peak_vals = signal_smooth[peaks]
    sort_idx = np.argsort(peak_vals)[::-1]

    #x strongest peaks where x = predicted age
    x = max(1, pred_age)
    chosen_idx = sort_idx[:x]
    chosen_peaks = peaks[chosen_idx]
    chosen_vals = peak_vals[chosen_idx]

    #Heatmap overlay = right now no ovrlay
    heatmap = plt.cm.jet(cam_norm)[..., :3]
    overlay = 1 * img_np + 0 * heatmap
    overlay = np.clip(overlay, 0, 1)

    #Plot
    fig = plt.figure(figsize=(15, 10))

    ax1 = fig.add_subplot(2, 2, 1)
    ax1.imshow(img_np)
    ax1.set_title("Originalbild")
    ax1.axis("off")
    ax2 = fig.add_subplot(2, 2, 2)
    ax2.imshow(overlay)
    ax2.scatter(chosen_peaks, [row] * len(chosen_peaks),
                color="yellow", s=40)
    ax2.axhline(row, color="white", linestyle="--", linewidth=1)
    ax2.set_title(f"Grad-CAM Overlay (valda {x} peaks)")
    ax2.axis("off")

    ax3 = fig.add_subplot(2, 1, 2)
    ax3.plot(signal_smooth, label="Grad-CAM längs rad")
    ax3.scatter(chosen_peaks, chosen_vals, color="red", label="valda peaks")
    ax3.set_xlabel("pixel (x)")
    ax3.set_ylabel("intensitet")
    ax3.legend()

    plt.tight_layout()
    plt.show()
```
