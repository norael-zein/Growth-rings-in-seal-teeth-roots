```python
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import os
import random
```

```python
os.makedirs('NewImages_BGR', exist_ok=True)
```

```python
paths = os.listdir('Nora')
im_path = paths[random.randint(0,30)]
im_path = im_path[0:-4] 
print(im_path)

f,a = plt.subplots(1,4)
a = a.ravel()
a[0].imshow(Image.open(f'Nora/{im_path}.jpg'))
for i in range(1,4):
    im = Image.open(f'NewImages/{im_path}_{i}.jpg')
    arr = np.array(im)
    arr = arr[:, :, ::-1]
    a[i].imshow(arr)
```

```python
for image_path in os.listdir("NewImages"):
    if image_path.lower().endswith(('.jpg')):
        im = Image.open(f'NewImages/{image_path}')
        arr = np.array(im)
        arr = arr[:, :, ::-1]
        out_path = f'NewImages_BGR/{image_path}'
        Image.fromarray(arr).save(out_path) 
```
