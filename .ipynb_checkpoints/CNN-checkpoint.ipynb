```python
import os, re, glob
import numpy as np
import pandas as pd
from pathlib import Path
from PIL import Image

import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
import torch.nn.functional as F

EXCEL_PATH = "Labels split images.xlsx"
BASE_DIR   = "NewImages"
BATCH_SIZE = 4
EPOCHS     = 10
LR         = 1e-3
DEVICE     = "cuda" if torch.cuda.is_available() else "cpu"

#Read file
df = pd.read_excel(EXCEL_PATH)

#Excel uses other dashes
def normalize_name(s: str) -> str:
    s = str(s).replace("–","-").replace("—","-")
    return re.sub(r"\s+"," ", s).strip()

def build_path(row):
    name = normalize_name(row["image_name_unlabelled"])
    base = str(Path(BASE_DIR) / name)
    hits = glob.glob(base + "_*.*")
    if not hits:
        hits = glob.glob(base + ".*")
    return hits[0] if hits else None

df["filepath"] = df.apply(build_path, axis=1)
df["age"] = pd.to_numeric(df["age"], errors="coerce")
df = df.dropna(subset=["filepath","age"]).reset_index(drop=True)

paths  = df["filepath"].values
labels = df["age"].astype("float32").values

if len(paths) == 0:
    raise RuntimeError("No matching files")

#80/20-split
idx = np.random.permutation(len(paths))
cut = int(0.8*len(paths))
train_paths, train_labels = paths[idx[:cut]], labels[idx[:cut]]
val_paths,   val_labels   = paths[idx[cut:]], labels[idx[cut:]]

#Dataset
class StripeDataset(Dataset):
    def __init__(self, paths, labels):
        self.paths = list(paths)
        self.labels = list(labels)
        self.tf = transforms.ToTensor()

    def __len__(self): return len(self.paths)

    def __getitem__(self, i):
        img = Image.open(self.paths[i]).convert("RGB")
        img = self.tf(img)                 # [C,H,W], float32 i [0,1]
        y   = torch.tensor(self.labels[i]) # skalar
        return img, y

def pad_collate(batch):
    imgs, labels = zip(*batch)
    H = max(im.shape[1] for im in imgs)
    W = max(im.shape[2] for im in imgs)
    padded = [F.pad(im, (0, W-im.shape[2], 0, H-im.shape[1])) for im in imgs]
    return torch.stack(padded, 0), torch.tensor(labels, dtype=torch.float32)

train_loader = DataLoader(StripeDataset(train_paths, train_labels),
                          batch_size=BATCH_SIZE, shuffle=True,
                          collate_fn=pad_collate)
val_loader   = DataLoader(StripeDataset(val_paths, val_labels),
                          batch_size=BATCH_SIZE, shuffle=False,
                          collate_fn=pad_collate)

#CNN
class CNN_model(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(3, 32, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
            nn.Conv2d(32, 64, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
            nn.Conv2d(64,128,3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
        )
        self.gap = nn.AdaptiveAvgPool2d((1,1))
        self.fc  = nn.Linear(128, 1)

    def forward(self, x):
        x = self.conv(x)
        x = self.gap(x).flatten(1)  
        return self.fc(x).squeeze(1)  

model = CNN_model().to(DEVICE)
opt = torch.optim.Adam(model.parameters(), lr=LR)
loss_fn = nn.L1Loss()
```

```python
hist_train_mae = []
hist_val_mae   = []

for epoch in range(1, EPOCHS+1):
    # ---- TRAIN ----
    model.train()
    train_mae, train_n = 0.0, 0
    for imgs, y in train_loader:
        imgs, y = imgs.to(DEVICE), y.to(DEVICE)
        opt.zero_grad()
        pred = model(imgs)
        loss = loss_fn(pred, y)  # MAE
        loss.backward()
        opt.step()
        train_mae += loss.item()
        train_n += 1
    train_mae /= max(1, train_n)

    # ---- VAL ----
    model.eval()
    val_mae, val_n = 0.0, 0
    with torch.no_grad():
        for imgs, y in val_loader:
            imgs, y = imgs.to(DEVICE), y.to(DEVICE)
            val_mae += loss_fn(model(imgs), y).item()
            val_n += 1
    val_mae /= max(1, val_n)

    # Spara historik
    hist_train_mae.append(train_mae)
    hist_val_mae.append(val_mae)

    print(f"Epoch {epoch}/{EPOCHS} | Train MAE: {train_mae:.3f} | Val MAE: {val_mae:.3f}")

#Save model
#torch.save(model.state_dict(), "stripe_regressor_pytorch.pth")
#print("Sparad: stripe_regressor_pytorch.pth")
```

```python
import matplotlib.pyplot as plt
import torch

epochs = range(1, len(hist_train_mae) + 1)

plt.figure(figsize=(6,4))
plt.plot(epochs, hist_train_mae, label="Train MAE")
plt.plot(epochs, hist_val_mae,   label="Val MAE")
plt.xlabel("Epoch")
plt.ylabel("MAE")
plt.title("MAE over epochs")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

model.eval()
y_true, y_pred = [], []
with torch.no_grad():
    for imgs, y in val_loader:
        preds = model(imgs.to(DEVICE)).cpu().numpy()
        y_pred.extend(preds.tolist())
        y_true.extend(y.numpy().tolist())

# 45°-linje
lo = min(min(y_true), min(y_pred))
hi = max(max(y_true), max(y_pred))

plt.figure(figsize=(5,5))
plt.scatter(y_true, y_pred, s=12)
plt.plot([lo, hi], [lo, hi], linestyle="--")
plt.xlabel("True age")
plt.ylabel("Predicted age")
plt.title("Pred vs. True (val)")
plt.grid(True)
plt.tight_layout()
plt.show()
```
